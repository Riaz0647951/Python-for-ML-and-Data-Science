```python
import matplotlib.pyplot as plt
import numpy as np
```

### Plot Customization in Matplotlib

```python
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

plt.title('Class Size', fontsize=25,
                        family='Arial',
                        fontweight='bold',
                        color='#2146ff')

plt.xlabel('Year', fontsize=20,
                   family='Arial',
                   fontweight='bold',
                   color='#2197ff')

plt.ylabel('Students', fontsize=20,
                       family='Arial',
                       fontweight='bold',
                       color='#2197ff')

# plt.tick_params() --> used to customize the values/ticks of x, y axes
plt.tick_params(axis='both',
                colors='#2197ff')

plt.plot(x, y, marker='*',
               markersize=15,
               markerfacecolor='cyan',
               markeredgecolor='green',
               #linestyle='solid'
               #linestyle='dashed'
               linestyle='dashdot',
               linewidth=2,
               color='purple')

plt.xticks(x) # it displays the x-axis values exactly as it is in the x-array

plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([28, 22, 32, 40])
y3 = np.array([13, 15, 20, 30])

line_style = dict(marker='*',
               markersize=15,
               markerfacecolor='cyan',
               markeredgecolor='green',
               #linestyle='solid'
               #linestyle='dashed'
               linestyle='dashdot',
               linewidth=2)


plt.title('Class Size', fontsize=25,
                        family='Arial',
                        fontweight='bold',
                        color='#2146ff')

plt.xlabel('Year', fontsize=20,
                   family='Arial',
                   fontweight='bold',
                   color='#2197ff')

plt.ylabel('Students', fontsize=20,
                       family='Arial',
                       fontweight='bold',
                       color='#2197ff')

# plt.tick_params() --> used to customize the values/ticks of x, y axes
plt.tick_params(axis='both',
                colors='#2197ff')

"""
The '**' before the 'line_style' unpack the dictionary --> it's as if we are passing 
the individual styles
"""
plt.plot(x, y1, color='blue', **line_style)
plt.plot(x, y2, color='purple', **line_style)
plt.plot(x, y3, color='brown', **line_style)

plt.xticks(x) # it displays the x-axis values exactly as it is in the x-array

plt.show()
```

### Grid lines

grid() = Helps make plots easier to read by adding reference lines

```python
x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.grid(axis='y',
         linewidth=2,
         color='lightgray',
         linestyle='dashed')

plt.plot(x, y)
plt.xticks(x)
plt.yticks(y)
plt.show()
```

### Bar Chart

Bar chart =  Compare categories of data by representing each category with a bar

```python
categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])

plt.bar(categories, values, color='skyblue')
# plt.barh(categories, values, color='skyblue') # plt.barh() --> horizontal bar chart

plt.title('Daily Consumption')
plt.xlabel('Food')
plt.ylabel('Quantity')

plt.show()
```

### Pie Chart

Pie chart = Circular chart divided into slices to show percentages of the total.

Good for visualizing distribution among categories.

```python
categories = np.array(["Freshmen", "Sophomores", "Juniors", "Seniors"])
values = np.array([300, 250, 275, 225])
colors = ["#64018f", "#02b6ed", "#3502ed", "#02ed9b"]
plt.pie(values, labels=categories,
                autopct='%1.1f%%',
                colors=colors,
                explode=[0, 0, 0, 0.1],
                shadow=True,
                startangle=90) # explode --> make slice a bit out of the pie chart

plt.title("College Student Categories")

plt.show()
```

### Scatter Plot

Scatter graph = Shows the relationship between two variables, \
Helps to identify a correlation (+, -, None) \
Example: Study hours vs. Test scores

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]) # Hours studied
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) # Grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) # Hours studied
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) # Grades

"""
alpha -->related to transparency
s --> size of data-points
"""
plt.scatter(x1, y1, color='skyblue',
                  alpha=0.7,
                  s = 200,
                  label='Class A')

plt.scatter(x2, y2, color='red',
                  alpha=0.7,
                  s = 200,
                  label='Class B')

plt.title('Test Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')

plt.legend()
plt.show()
```

### Histogram

Histogram = A visual representation of the distribution of quantitative data. \
They group values into bins (intervals) and counts how many fall in each range.

```python
scores = np.random.normal(loc=80, scale=10, size=100)
scores = np.clip(scores, 0, 100) # .clip() --> to clip the values below 0 and above 100

plt.hist(scores, bins=10,
                 color='lightgreen',
                 edgecolor='black')
plt.title('Exam Scores')
plt.xlabel('Score')
plt.ylabel('# of students')
plt.show()
```

### Subplots

Figure = The entire canvas \
Ax = A single plot (subplot)

```python
x = np.array([1, 2, 3, 4, 5])

figure, axes = plt.subplots(2,2)
axes[0,0].plot(x, x*2, color='red')
axes[0,0].set_title('x*2')

axes[0,1].plot(x, x**2, color='blue')
axes[0,1].set_title('x**2')

axes[1,0].plot(x, x**3, color='green')
axes[1,0].set_title('x**3')

axes[1,1].plot(x, x**4, color='purple')
axes[1,1].set_title('x**4')

plt.tight_layout()
plt.show()
```
