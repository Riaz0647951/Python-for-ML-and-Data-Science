```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'D:\Python-for-Data-Science-&-ML\Matplotlib\dataset\data.csv')
df
```

```python
# Categories in Type 1 column and their value counts
Type1_counts = df['Type1'].value_counts(ascending=True)
Type1_counts
```

```python
# Bar chart
plt.barh(Type1_counts.index, Type1_counts.values, color='#03dffc',
                                                  edgecolor='black')
plt.title('# of Pokemon by Primary Type')
plt.xlabel('Count')
plt.ylabel('Type')
plt.tight_layout()
plt.show
```
