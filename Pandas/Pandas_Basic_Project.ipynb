```python
df = pd.read_csv(r'D:\Python-for-Data-Science-&-ML\Pandas\Datasets\data.csv', index_col='Name')

# Selection By Row using location(label)
df.loc['Pikachu']
```

```python
# Selection By Row using location(label)
df.loc['Charizard', ['Height', 'Weight']]
```

```python
# Selection By Rows using location(label)
# Get everything between 'Charizard' and 'Blastoise'
df.loc['Charizard':'Blastoise', ['Height', 'Weight']]
```

```python
# Selection By Rows using integer location
df.iloc[0:11]
```

```python
# Selection By Rows using integer location
# Get every second row between 0 to 10 rows, and the first 3 columns
df.iloc[0:11:2, 0:3]
```

```python
pokemon = input('Enter a Pokemon name: ')

try:
    print(df.loc[pokemon])
    
except KeyError:
    print(f'{pokemon} not found!')
```

```python
# Filtering = Keeping the rows that match a condition
# Get the pokemon having 'Height' greater than or equal to 2
tall_pokemon = df[df['Height'] >= 2]

print(tall_pokemon)
```

```python
# Return the pokemon having 'Weight' greater than 100
heavy_pokemon = df[df['Weight'] > 100]

print(heavy_pokemon)
```

```python
# Return the pokemon that is legendary (Legendary == 1)
legendary_pokemon = df[df['Legendary'] == 1]

print(legendary_pokemon)
```

```python
# Lets find the water pokemon
water_pokemon = df[df['Type1'] == 'Water']

print(water_pokemon)
```

```python
# Find the pokemon having either Type1 or Type2 == 'Water'
water_pokemon = df[(df['Type1'] == 'Water') | (df['Type2'] == 'Water')]

print(water_pokemon)
```

```python
# Lets find pokemon that are 'Fire' and 'Flying'
ff_pokemon = df[(df['Type1'] == 'Fire') & (df['Type2'] == 'Flying')]
print(ff_pokemon)
```

```python
"""
Aggregate functions = Reduce a set of values into a single summary value
Used to summarize and analyze data
Often used with the groupby() function
"""
# Applied to whole DataFrame
print("mean of all numeric columns: \n", df.mean(numeric_only=True))

print("\nsum of all numeric columns: \n", df.sum(numeric_only=True))

print("\nminimum of all numeric columns: \n", df.min(numeric_only=True))

print("\nmaximum of all numeric columns: \n", df.max(numeric_only=True))

print("\ncount the number of values within each column: \n", df.count())
```

```python
# Single Column

print("mean of the 'Height' column: \n", df['Height'].mean())

print("\nsum of all the values of 'Height' column: \n", df['Height'].sum())

print("\nminimum value in the 'Height' column: \n", df['Height'].min())

print("\nmaximum value in the 'Height' column: \n", df['Height'].max())

print("\ncount the number of values within 'Height' column: \n", df['Height'].count())
```

```python
# groupby()
group = df.groupby("Type1")

# Return the average height for each group of pokemon
print("The average height for each group of pokemon: \n", group['Height'].mean())

# Return the sum of height for each group of pokemon
print("\nThe sum of height for each group of pokemon: \n", group['Height'].sum())
```

```python
# Return the minimum height for each group of pokemon
print("\nThe minimum height for each group of pokemon: \n", group['Height'].min())
```

```python
# Return the maximum height for each group of pokemon
print("\nThe maximum height for each group of pokemon: \n", group['Height'].max())
```

```python
# Return the count of height for each group of pokemon
print("\nThe count of height for each group of pokemon: \n", group['Height'].count())
```
