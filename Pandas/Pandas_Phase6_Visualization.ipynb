##  Pandas Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'D:\Python-for-Data-Science-&-ML\Pandas\Datasets\Ice Cream Ratings.csv')
df
```

```python
df = df.set_index('Date')
df
```

```python
#df.plot() --> by default it give us a line chart/plot, but we can specify the kind of chart
df.plot()
```

```python
df.plot(kind='line')
```

```python
# We can also plot subplots
df.plot(kind='line', subplots=True)
```

```python
print(plt.style.available) # This prints the available color styles in matplotlib
```

```python
# We can use any available style from the above list for our plots
plt.style.use('classic') # Uses a style(classic) from the above list
```

```python
# We can customize the plot by using title, xlabel,and  ylabel as per our wish/requirement
df.plot(kind='line', title= 'Ice Cream Ratings', xlabel='Daily Ratings', ylabel='Scores')
```

```python
# Bar Chart
df.plot(kind='bar')
```

```python
# Stacked Bar Chart
df.plot(kind='bar', stacked=True)
```

```python
# We can also plot a specific columns
df['Flavor Rating'].plot(kind='bar')
```

```python
# We can also plot horizontal bar chart
df.plot.barh()
```

```python
# Horizontal Stacked bar chart
df.plot.barh(stacked=True)
```

```python
"""
We can also plot scatter plot, but for this we have to specify x and y axis
in order to run scatter plot properly.
s = size of points
c = color
"""
df.plot.scatter(x = 'Texture Rating', y = 'Overall Rating', s=200, c='orange')
```

```python
# Histogram plot --> by default it generate 10 bins, but we can specify no. of bins
df.plot.hist()
```

```python
# Histogram plot --> by default it generate 10 bins, but we can specify no. of bins
df.plot.hist(bins=20)
```

```python
# Boxplot
df.boxplot()
```

```python
# Area plot
# figsize --> to adjust the size of figure (horizontal length, vertical width)
df.plot.area(figsize=(10,5))
```

```python
 # Pie chart --> it requires either y column or subplots=True
 df.plot.pie(y='Flavor Rating')
```
