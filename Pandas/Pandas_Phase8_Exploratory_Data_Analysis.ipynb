## Exploratory Data Analysis (EDA) in Pandas

EDA is a first look at data, identifying patterns within data, understanding the relationships between features, looking at outliers that may exist within the dataset, and also looking for mistakes and missing values.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'D:\Python-for-Data-Science-&-ML\Pandas\Datasets\world_population.csv')
df
```

```python
# The population values above are upto 1 decimal place, lets make it upto 2 decimal
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

```python
df
```

```python
# Get all the information about the dataset
df.info()
```

```python
# Get statistical summary of the dataset
df.describe()
```

```python
# Get the count of null values in all the columns
df.isnull().sum()
```

```python
# Get how many unique values are there in each of the columns
df.nunique()
```

```python
"""
sort the dataset by "2022 Population" in ascending order and return the top 5 rows means --> 
return the top 5 countries having the lowest populations in 2022
"""
df.sort_values(by='2022 Population').head()
```

```python
"""
sort the dataset by "2022 Population" in descending order and return the top 10 rows means --> 
return the top 10 countries having the highest populations in 2022
"""
df.sort_values(by='2022 Population', ascending=False).head(10)
```

```python
"""
sort the dataset by "World Population Percentage" in descending order and return top 10 rows 
--> return top 10 countries that contributes the largest proportion to the world population
"""
df.sort_values(by='World Population Percentage', ascending=False).head(10)
```

```python
# Look into correlation between columns with numeric values only
df.corr(numeric_only= True)
```

```python
# Visualize the correlation
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show
```

```python
# Group the dataset by "Continent" and return their means sorted by "2022 Population" in descending order 
df.groupby('Continent').mean(numeric_only=True).sort_values(by='2022 Population', ascending=False)
```

```python
# Get the dataframe that contains the name "Oceania" in the "Continent" column
df[df['Continent'].str.contains('Oceania')]
```

```python
"""
 Group the dataset by 'Continent' only for populations columns and return their means 
 sorted by "2022 Population" in descending order 
"""

df2 = df.groupby('Continent')[['1970 Population',
       '1980 Population', '1990 Population', '2000 Population',
       '2010 Population', '2015 Population', '2020 Population',
       '2022 Population']].mean(numeric_only=True).sort_values(by='2022 Population', ascending=False)
df2
```

```python
# Take transpose
df3 = df2.transpose()
df3
```

```python
df3.plot(figsize=(15, 6))
```

```python
# Boxplot is really good for finding outliers
df.boxplot(figsize=(20,10))
```
