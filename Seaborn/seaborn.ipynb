```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

### Line Plot

```python
var = [1,2,3,4,5,6,7]
var_1 = [2,3,4,1,5,6,7]

# seaborn work on DataFrame, so we have to convert our data into DataFrame
x_1 = pd.DataFrame({"var":var, "var_1":var_1})
```

```python
# Line plot in seaborn
sns.lineplot(data=x_1, x="var", y="var_1")
plt.show()
```

```python
# we can load any dataset directly from github just be putting the name of dataset
y_1 = sns.load_dataset("penguins").head(20)
y_1
```

```python
sns.lineplot(data=y_1, x="bill_length_mm", y="bill_depth_mm", hue="sex", style="sex", 
             palette="Accent", markers=["o", ">"], dashes=False, legend="full")
plt.title("Python Practice")
plt.grid()
plt.show()
```

### Bar Plot

```python
var = sns.load_dataset("penguins")
var
```

```python
sns.barplot(x="island", y="bill_length_mm", data=var)
plt.show()
```

```python
order_1 = ["Dream", "Torgersen", "Biscoe"]
sns.set(style='darkgrid')
sns.barplot(x="island", y="bill_length_mm", data=var, hue="sex", order=order_1, 
            hue_order=["Female", "Male"], palette="icefire", saturation=0.75,
            err_kws={'color':'r', 'linewidth':5}, alpha=0.9)
plt.show()
```

```python
# Horizontal Bar plot
sns.barplot(x="bill_length_mm", y="island", data=var, hue="sex", order=order_1, 
            hue_order=["Female", "Male"], orient="h")
plt.show()
```

### Histogram Plot

```python
# Un-organized Histogram
sns.displot(var["flipper_length_mm"])
plt.show()
```

```python
# Organized Histogram
sns.displot(var["flipper_length_mm"], bins=[170, 180, 190, 200, 210, 220, 230, 240], 
            kde=True, rug=True, color=(0.6, 0.15, 0.96, 0.9))
plt.show()
```

### Scatter Plot

```python

sns.scatterplot(x="bill_length_mm", y="bill_depth_mm", data=var, hue="sex", style="sex",
                size="sex", sizes=[40,55], palette="gist_ncar", alpha=0.8)
plt.show()
```

### Heatmap Plot

```python
import numpy as np
var = np.linspace(1,10,20).reshape(4,5)
var
```

```python
y = {"fontsize":15, "color":"brown"}
v = sns.heatmap(var, vmin=0, vmax=10, cmap='PuOr', annot=True, annot_kws=y, 
            linewidths=4, linecolor="purple")
v.set(xlabel="Python", ylabel="Practice")
sns.set(font_scale=1)
plt.show()
```

```python
# load dataset (anagrams.csv) from github
data = sns.load_dataset('anagrams')
data
```

```python
# drop the string column (attnr), as heatmap requires only numeric data
x = data.drop(columns=["attnr"], axis=1).head(10)
x
```

```python
y = {"fontsize":10, "color":"green"}
sns.heatmap(x, annot=True, annot_kws=y, linewidths=2, linecolor="orange", cbar=False)
plt.show()
```

### Count Plot

```python
# load dataset (tips.csv) from github
var = sns.load_dataset("tips")
var
```

```python
# Vertical Count Plot
sns.countplot(data=var, x="sex", hue="smoker", palette="viridis")
plt.show()
```

```python
# Horizontal Count Plot
sns.countplot(data=var, y="sex", hue="smoker", palette="mako")
plt.show()
```

### Violin Plot

```python
sns.violinplot(data=var, x="day", y="total_bill", linewidth=2, color="orange")
plt.show()
```

```python
sns.violinplot(data=var, x="day", y="total_bill", linewidth=2, hue="time", palette="Dark2")
plt.show()
```

```python
sns.violinplot(data=var, x="time", y="total_bill", order=["Dinner", "Lunch"], 
               hue="time", palette="gist_rainbow", inner="quart")
plt.show()
```

```python
sns.violinplot(x="day", y="total_bill", data=var, hue="sex", split=True)
plt.show()
```

```python
sns.violinplot(x="total_bill", y="day", data=var, hue="sex", split=True)
plt.show()
```

### Pair Plot

```python
sns.pairplot(var, hue="sex")
plt.show()
```

```python
sns.pairplot(var, vars=["total_bill", "tip"], hue="sex")
plt.show()
```

```python
sns.pairplot(var, hue="sex", x_vars=["total_bill", "tip"], palette="gist_rainbow")
plt.show()
```

```python
sns.pairplot(var, hue="sex", y_vars=["total_bill", "tip"], palette="rainbow")
plt.show()
```

```python
sns.pairplot(var, hue="sex", kind="reg", diag_kind="hist") # kind = "reg", "scatter", "kde", "hist"
plt.show()
```

### Strip Plot

```python
sns.stripplot(x="day", y="total_bill", data=var, hue="sex", palette="rocket_r")
plt.show()
```

```python
sns.stripplot(x="day", y="total_bill", data=var, hue="sex", palette="rocket_r", 
              linewidth=1, edgecolor="green", size=7)
plt.show()
```

### Box Plot

```python
sns.set(style="whitegrid")
sns.boxplot(x="day", y="total_bill", data=var, palette='plasma', showmeans=True, 
            meanprops={"marker":"+", "markeredgecolor":"red"}, linewidth=2)
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="day", y="total_bill", data=var, hue="time")
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="day", y="total_bill", data=var, hue="sex", palette='light:green', 
            showmeans=True, meanprops={"marker":"+", "markeredgecolor":"red"}, 
            linewidth=1)
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="total_bill", y="day", data=var, palette='plasma', showmeans=True, 
            meanprops={"marker":"+", "markeredgecolor":"red"}, linewidth=2)
plt.show()
```

### Multiple Plots (Facet-Grid)

```python
fg = sns.FacetGrid(var, col="sex", hue="day")
fg.map(plt.scatter, "tip", "total_bill").add_legend()
plt.show()
```

```python
fg = sns.FacetGrid(var, col="day", hue="sex")
fg.map(plt.scatter, "tip", "total_bill").add_legend()
plt.show()
```

```python
fg = sns.FacetGrid(var, col="day", hue="sex", palette="summer")
fg.map(plt.bar, "tip", "total_bill").add_legend()
plt.show()
```
