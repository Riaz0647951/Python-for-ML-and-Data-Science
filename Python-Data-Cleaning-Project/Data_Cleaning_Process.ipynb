### Handling Missing Data

```python
df.dropna(subset=['Customer_ID'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Age'].unique()
```

```python
df['Age'] = df['Age'].str.replace(' years', '', regex=False)
```

```python
df['Age'].unique()
```

```python
df_age = df[df['Age'] != 'nan']['Age']
```

```python
df_age = pd.to_numeric(df_age, errors='coerce').astype('Int64')
```

```python
age_median = int(df_age.dropna().astype('int64').median())
age_median
```

```python
df['Age'].unique()
```

```python
df['Age'] = df['Age'].replace(to_replace='nan', value=age_median)
```

```python
df['Age'].unique()
```

```python
df['Age'].fillna(age_median, inplace=True)
```

```python
df['Age'].unique()
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce').astype('Int64')
```

```python
df['Age'].unique()
```

```python
df['Age'] = df['Age'].astype('object')
```

```python
df['Age'].unique()
```

```python
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(df['Purchase_Amount'].median())
df['Purchase_Amount']
```

```python
df.isnull().sum()
```

```python
df['Feedback_Score'] = df['Feedback_Score'].fillna(df['Feedback_Score'].mode()[0])
```

```python
df.isnull().sum()
```

```python
for col in ['Gender', 'City', 'Country']:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
# forward fill
df['Last_Purchase_Date'].ffill(inplace=True) 
```

```python
df.isnull().sum()
```

```python
import seaborn as sns
# Boxplot -- To highlight outliers
cols = ['Age', 'Purchase_Amount', 'Feedback_Score']

plt.figure(figsize=(15, 5))

for i, col in enumerate(cols, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[col], color='skyblue', width=0.3)
    plt.title(f'Boxplot of {col}', fontsize=12)
    plt.ylabel('')
    plt.grid(axis='y', linestyle='--', alpha=0.6)
    
plt.tight_layout()
plt.show()
```

```python
from scipy import stats
import numpy as np
```

```python
# calculating z-score
z_scores = np.abs(stats.zscore(df[['Age', 'Purchase_Amount']].astype(float)))
```

```python
df_clean = df[~(z_scores > 3).any(axis = 1)]
```

```python
df_clean.shape
```

```python
# Boxplot of clean data to check whether the outliers are removed

cols = ['Age', 'Purchase_Amount', 'Feedback_Score']
  
plt.figure(figsize=(15, 5))

for i, col in enumerate(cols, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df_clean[col], color='skyblue', width=0.3)
    plt.title(f'Boxplot of {col}', fontsize=12)
    plt.ylabel('')
    plt.grid(axis='y', linestyle='--', alpha=0.6)
    
plt.tight_layout()
plt.show()
```
