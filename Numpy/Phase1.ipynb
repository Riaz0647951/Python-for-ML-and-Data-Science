## Numpy Foundations

```python
import numpy as np
```

### Creating array from list

```python
arr_1d = np.array([1, 2, 3])
print("1D array: " ,arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array: " ,arr_2d)
```

### List vs numpy array

```python
py_list = [1, 2, 3]
print("Python list multiplication: ", py_list * 2)


np_array = np.array([1, 2, 3]) # element-wise multiplication
print("Python numpy array multiplication: ", np_array * 2)
```

```python
import time
start = time.time()
py_List = [i*2 for i in range(1000000)]
print("\n List Operation Time: ", time.time() - start)

start = time.time()
np_Array = np.arange(1000000) * 2
print("\n Numpy Operation Time: ", time.time() - start)
```

### Creating array from scratch

```python
zeros = np.zeros((3, 4)) # create 3x4 matrix/array of zeros
print("zeroes array: \n", zeros)

ones = np.ones((2, 3)) # create 2x3 matrix/array of ones
print("ones array: \n", ones)

full = np.full((3,4), 5) # create 3x4 matrix/array of constant number 5
print("full array: \n", full)

random = np.random.random((2,3)) # create random array. It create a 2x3 matrix of values b/w 0 and 1
print("random array: \n", random)

sequence = np.arange(0, 10, 2) # create a squence of even numbers from 0 to 10(exclusive).
print("sequence array: \n", sequence)

evenly_space_array = np.linspace(0, 1, 10) # create array of 5 evenly space numbers between 0 and 1 (start, stop, number of values)
print('Evenly spaced array: \n', evenly_space_array)
```

```python
# Logrithmic scale array --> 10^1 to 10^3, and total 3 values b/w them
log_array = np.logspace(1, 3, 3) # (Starting, ending --> powers, No. of values)
print('Logarithmic scale array: \n', log_array)
```

```python

# randn --> random floats from standard normal distribution
random_standard_array = np.random.randn(2,3) 
print('random normal distribution array: \n', random_standard_array)
```

```python
# randint --> random integers
rand_int = np.random.randint(1, 50, size = (4,4)) # start, stop, no. of values
print('Random integers: \n', rand_int)
```

```python
# Identity matrix
identity_matrix = np.eye(3) # identity matrix with 3 rows and 3 columns
print('Identity matrix: \n', identity_matrix)
```

### Vector, Matrix, and Tensor

```python
vector = np.array([1, 2, 3]) # 1-D array is a Vector
print("Vector: ", vector)

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]]) # 2-D array is called Matrix
print("Matrix: ", matrix) 

tensor = np.array([[[1, 2], [3, 4]],
                   [[5, 6], [7, 8]]]) # Array with 3 or more dimensions/multi-dimensional array is called Tensor
print("Tensor: ", tensor)
```

### Array Properties

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print("Shape: ", arr.shape) # shape shows number of rows & columns
print("Dimension: ", arr.ndim) # dimension of array like 1-D, 2-D etc...
print("Size: ", arr.size) # size shows the total number of elements in an array
print("Data Type: ", arr.dtype) # It tells about the data type like int, string etc...
```

### Array Reshaping

```python
arr = np.arange(12) # array of numbers from 0 to 11
print("Original array: \n", arr)

reshaped = arr.reshape((3, 4)) # convert theoriginal array to 3x4 array/matrix
print("\nReshaped array: \n", reshaped)

flattened = reshaped.flatten() # Convert multi-dimensional array to flat 1-D array
print("\nFlattened array: \n", flattened)

# ravel (returns view, instead of copy like flatten)
raveled = reshaped.ravel() # It also converts multi-dimensional array to flat 1-D array
print("\nRaveled array: \n", raveled)

transpose = reshaped.T # take transpose --> convert rows into columns and vice versa
print("\nTranspose array: \n", transpose)

transposed = reshaped.transpose() # it also convert rows into columns and vice versa
print("\nTransposed array: \n", transposed)
```

```python
# swapaxes --> swap 2 specific axes in a matrix
arr = np.array([[[1, 2], [3, 4]]])
print('shape: ', arr.shape)

swap = np.swapaxes(arr, 1, 2) # array, 1st axis, 2nd axis
print('swap shape: ', swap.shape)

print('swap array: \n', swap)
```

```python
# Splitting array
arr = np.array([[1, 2], [3, 4]])
print('split array: \n', np.split(arr, 2)) # Split original array into 2 parts/arrays

print('horizontal split array: \n', np.hsplit(arr, 2)) # Split original array into 2 parts/arrays column wise

print('vertical split array: \n', np.vsplit(arr, 2)) # Split original array into 2 parts/arrays row wise
```

```python
# Repeating
array = np.array([1, 2, 3])
print(np.repeat(array, 2)) # repeat each element of array twice
```

```python
# Tile --> Repeat whole array
print(np.tile(array, 2)) # repeat whole array twice
```
