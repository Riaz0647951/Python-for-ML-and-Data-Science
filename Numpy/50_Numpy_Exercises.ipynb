### 44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates

```python
np.random.seed(42)
matrix = np.random.randint(0,10,(10,2))
print("Cartesian Coordinates Matrix: \n", matrix)
x = matrix[:,0]
y = matrix[:,1]

"""
polar coordinates = (r, theta)
r = (x^2  + y^2) ^ (1/2)
theta = arctan(y/x)
"""

r = np.hypot(x, y)
theta = np.arctan2(y, x)

polar_rad = np.column_stack((r,theta))
print("Polar Coordinates Matrix (Radian): \n", polar_rad)

polar_deg = np.column_stack((r,np.degrees(theta)))
print("Polar Coordinates Matrix (Degree): \n", polar_deg)
```

### 47. Given two arrays X and Y, construct the cauchy matrix C (Cij = 1/(xi - yj))

```python
X = np.array([6,7,8,9,10])
Y = np.array([1,2,3])

C = np.zeros((len(X), len(Y)))
for i in range(len(X)):
    for j in range(len(Y)):
        C[i,j] = 1/(X[i]-Y[j])
        
C
```

### 50. How to find the closest value (to a given scalar) in a vector?

```python
z = np.arange(100)
v = np.random.uniform(0,100)
index = (np.abs(z-v)).argmin()
print("z --> array: \n", z)
print("v --> uniform array: \n", v)
print("Closest value: \n",z[index])
```
